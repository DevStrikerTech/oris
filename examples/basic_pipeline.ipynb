# Basic Oris pipeline

Load the sample YAML under `examples/`, run it with `Pipeline.run`, and inspect the **output** and **trace summary**.

**Tip:** Run Jupyter with the working directory set to the **repository root** or to **`examples/`** so the helper below can find `simple_generation.yaml`.

```python
from pathlib import Path


def examples_dir() -> Path:
    here = Path.cwd().resolve()
    if (here / "simple_generation.yaml").exists():
        return here
    ex = here / "examples"
    if (ex / "simple_generation.yaml").exists():
        return ex
    raise FileNotFoundError(
        "Could not find simple_generation.yaml — open Jupyter from repo root or examples/."
    )


EX = examples_dir()
EX
```

```python
from pprint import pprint

from oris import Pipeline

pipeline = Pipeline.from_yaml(EX / "simple_generation.yaml")
result = pipeline.run({"query": "What is responsible AI?"})

print("--- result.output ---")
pprint(result.output)
```

```python
import json

summary = result.to_run_summary()
print("--- to_run_summary() (structured CLI-style view) ---")
print(json.dumps(summary, indent=2, default=str))
```

## Trace / debug-style fields

Each entry in `summary["trace"]` includes `step_id`, `component_name`, `status`, `latency_ms`, and `flags`. Use `oris run ... --debug` on the CLI to print similar step lines to **stderr** while keeping JSON on stdout.
